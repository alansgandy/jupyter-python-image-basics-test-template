```python
# Import necessary libraries
import matplotlib; matplotlib.use('Agg')
import matplotlib.pyplot as plt
import numpy as np
import cv2
```

## Task 1: Load an Image
Implement the function to load an image from a file path. Use OpenCV to read the image.

```python
def load_image(file_path):
    # TODO: Load the image using cv2.imread
    # 1. Use cv2.imread to read the image from file_path
    # 2. Return the loaded image
    image = cv2.imread(file_path)
    return image
```

## Task 2: Display an Image
Implement the function to display an image using matplotlib.

```python
def display_image(image):
    # TODO: Display the image using matplotlib
    # 1. Use plt.imshow to display the image
    # 2. Convert BGR to RGB if necessary
    # 3. Use plt.axis('off') to hide axes
    # 4. Use plt.close() instead of plt.show()
    if len(image.shape) == 3:  # Check if the image is colored
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    plt.imshow(image, cmap='gray')
    plt.axis('off')
    plt.show()
    plt.close()
```

## Task 3: Convert to Grayscale
Implement the function to convert a color image to grayscale.

```python
def convert_to_grayscale(image):
    # TODO: Convert the image to grayscale
    # 1. Use cv2.cvtColor with cv2.COLOR_BGR2GRAY
    # 2. Return the grayscale image
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    return gray_image
```

## Task 4: Main Block
This block will use the functions you implemented to load, display, and convert an image.

```python
# Main block
# Hardcoded data: Use a sample image path
image_path = 'sample_image.jpg'

# Load the image
image = load_image(image_path)

# Display the original image
display_image(image)

# Convert the image to grayscale
gray_image = convert_to_grayscale(image)

# Display the grayscale image
display_image(gray_image)
```
